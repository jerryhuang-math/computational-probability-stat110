# birthday problem
There are k people in a room. Assume each person’s birthday is equally likely to be any of the 365 days of the year (we exclude February 29), and that people’s birthdays are independent (we will define indepen dence formally later, but intuitively it means that knowing some people’s birthdays gives us no information about other people’s birthdays; this would not hold if, e.g., we knew that two of the people were twins). What is the probability that at least one pair of people in the group have the same birthday?

```python
import time
import numpy as np

start_time = time.perf_counter()

N = 23

stimulation = 1000000
count = 0

for k in range(stimulation):
    ind = 0
    birth = np.random.choice(365, size=N, replace=True)
    for i in range(N):
        for j in range(i+1,N):
            if birth[i] == birth[j]:
                count += 1
                ind = 1
                break
        if ind == 1:
            break

prob = count/stimulation
print(prob)

end_time = time.perf_counter()
execution_time = end_time - start_time
print(f"Running time: {execution_time:.6f} seconds")
```

```python
# Second try

start_time = time.perf_counter()

N = 23

stimulation = 1000000
count = 0

for k in range(stimulation):
    birth = np.random.randint(0, 365, size=N)
    birth_set = set(birth)
    if len(birth) != len(birth_set):
        count += 1

prob = count/stimulation
print(prob)

end_time = time.perf_counter()
execution_time = end_time - start_time
print(f"Running time: {execution_time:.6f} seconds")
```

```python
# Third try

start_time = time.perf_counter()

N = 23
stimulation = 1000000
exper_set = np.random.randint(0, 366, size= N *stimulation).reshape(stimulation, -1)

exper_set.sort(axis=1)
detect_zero = exper_set[:, 1:]- exper_set[:, :-1]
prob = sum(np.any(detect_zero == 0, axis=1)) / stimulation

print(prob)

end_time = time.perf_counter()
execution_time = end_time - start_time
print(f"Running time: {execution_time:.6f} seconds")
```

# de Montmort’s matching problem
Consider a well-shuffled deck of n cards, labeled 1 through n. You flip over the cards one by one, saying the numbers 1 through n as you do so. You win the game if, at some point, the number you say aloud is the same as the number on the card being flipped over (for example, if the 7th card in the deck has the label 7). What is the probability of winning?

```python
import numpy as np

n = 52
s = 1000000
ordered_set = np.tile(np.arange(1, n + 1), (s, 1))
rows = np.arange(ordered_set.shape[0])[:, None]
index = np.argsort(np.random.rand(*ordered_set.shape), axis=1)
shuffled_set = ordered_set[rows,index]
detect_zero = shuffled_set - ordered_set
prob = np.mean(np.any(detect_zero == 0, axis =1))

print('Calculated probability', prob)
theo_value = 1- np.exp(-1)
print('It should approach', theo_value)
```
