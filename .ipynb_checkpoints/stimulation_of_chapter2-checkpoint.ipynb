```python
# The three door problem

import numpy as np

stimulation = 100000000
car_location = np.random.randint(0, 3, size = stimulation)
cont_choice = np.random.randint(0, 3, size = stimulation)

# Don't switch
prob_noswitch = np.mean((car_location - cont_choice) == 0)
print(prob_noswitch)

# Switch
switch_choice = cont_choice.copy()
switch_choice[car_location == cont_choice] = (switch_choice[car_location == cont_choice] + np.random.randint(1,3)) % 3
switch_choice[car_location != cont_choice] = car_location[car_location != cont_choice]
prob_switch = np.mean((car_location - switch_choice) == 0)
print(prob_switch)
```
